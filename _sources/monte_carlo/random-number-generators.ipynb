# Random Number Generators

A random number generator returns a sequence of numbers that should have no correlation from one to the next.  In particular, we want it to:

* Be fast
* Have a long period before repeating
* Uniformly samples the domain, with little correlation between adjacent numbers

In practice we use [pseudorandom number generators](https://en.wikipedia.org/wiki/Pseudorandom_number_generator).  These are implemented in software via an algorithm that gives us the sequence of numbers.

[True random number generators](https://en.wikipedia.org/wiki/Hardware_random_number_generator) are often very slow, and impractical.

## Linear congruential method

A simple pseudorandom number generator is the [linear congruential method](https://en.wikipedia.org/wiki/Linear_congruential_generator).  This takes the form of a recurrence relation:

$$x_{n+1} = (a x_n + c) \mbox{mod} m$$

where:

* $m > 0$ is the modulus which specifies the period of the sequence
* $0 < a < m$ is the multiplier
* $0 \le c < m$ is the increment

It starts with a *seed*, $x_0$.

The random number is then:

$$r = \frac{x_{n+1}}{M}$$

which falls in the range $[0, 1)$.



```{note}
A seed is a common concept in pseudorandom number generators.  For a particular seed, you get a specific sequence of random numbers.  This can allow for reproducibility with randomness, which can be important for testing.
```
