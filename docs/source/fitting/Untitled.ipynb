```python
from sympy import init_session
init_session()
```

```python
c2 = symbols("\chi^2")
a0, a1 = symbols("a_0 a_1")
xi, yi, si = symbols("x_i y_i \sigma_i")
```

```python
c2 = (yi - a0 * exp(a1 * xi))**2 / si**2
c2
```

```python
g0 = c2.diff(a0) / (-2)
g0
```

```python
g1 = c2.diff(a1) / (-2)
g1
```

```python
g0.diff(a0)
```

```python
simplify(g0.diff(a1))
```

```python
simplify(g1.diff(a0))
```

```python
simplify(g1.diff(a1))
```

```python
mui, z, dmui = symbols("\mu_i z \Delta{}\mu_i")
```

```python
c2 = (mui - 5 * log(a0 * z * (1 + Rational(1, 2) * (1 - a1) * z), 10) - 25)**2 / dmui**2
c2
```

```python
simplify(c2.diff(a0).diff(a0))
```
