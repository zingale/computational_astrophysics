```python
def find_primes(n):
    if n < 2:
        return []
    primes = [2]
    for num in range(3, n+1):
        is_prime=True
        for prime in primes:
            if num % prime == 0:
                is_prime=False
                break
        if is_prime:
            primes.append(num)
    return primes
```

```python
a  =find_primes(100)
print(a)
```
